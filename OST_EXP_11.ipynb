```python
import pandas as pd
import matplotlib.pyplot as plt

# Sample dataset
data = {
    "Subject": ["Maths", "Science", "English"],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

# Plot graph
df.plot(x="Subject", y="Marks", kind="bar")
plt.title("Marks Analysis")
plt.show()
```
